```python
# Importing the main libraries needed for this NLP assignment.
# NLTK is used for tokenizing, stop words, stemming, lemmatizing,
# named entity recognition, and n-grams.
import nltk
import string
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag, ne_chunk
from nltk.util import ngrams
from collections import Counter

nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")
nltk.download("averaged_perceptron_tagger")
nltk.download("averaged_perceptron_tagger_eng")
nltk.download("maxent_ne_chunker")
nltk.download("maxent_ne_chunker_tab")
nltk.download("words")
```

```python
# This function opens a text file, reads all of its content,
# closes the file, and returns the text as one string.
# I used utf-8-sig to avoid weird hidden characters at the start of the file.
def load_text(filename):
    file = open(filename, "r", encoding="utf-8-sig")
    text = file.read()
    file.close()
    return text
```

```python
# This dictionary stores the names of the four texts and the file names.
# This makes it easier to loop through all the files instead of repeating code.
texts = {
    "Text_1_RJ_Lovecraft": "RJ_Lovecraft.txt",
    "Text_2_RJ_Tolkien": "RJ_Tolkein.txt",
    "Text_3_RJ_Martin": "RJ_Martin.txt",
    "Text_4_Unknown_Martin": "Martin.txt"
}

# Creating the stop word list so common words like "the", "and", and "is"
# can be removed before analyzing word frequency.
stop_words = set(stopwords.words("english"))

# Creating the stemmer and lemmatizer objects.
# The stemmer reduces words to shorter stems, while the lemmatizer
# tries to reduce words to a more dictionary-like form.
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
```

```python
# This function cleans and prepares the text for analysis.
# It first tokenizes the text into words, then lowercases each token.
# It removes punctuation, numbers, and stop words.
# After that, it applies stemming and lemmatization before saving the token.

def preprocess_text(text):
    tokens = word_tokenize(text)

    filtered_tokens = []

    for token in tokens:
        token = token.lower()

        if token.isalpha() and token not in stop_words:
            stemmed = stemmer.stem(token)
            lemmatized = lemmatizer.lemmatize(stemmed)
            filtered_tokens.append(lemmatized)

    return filtered_tokens
```

```python
# This function finds named entities in a text.
# It tokenizes the text, assigns part-of-speech tags, and then uses ne_chunk
# to identify named entities such as people, places, and organizations.
def count_named_entities(text):
    tokens = word_tokenize(text)
    tagged_tokens = pos_tag(tokens)
    tree = ne_chunk(tagged_tokens)

    named_entities = []

    # Loop through the chunk tree and collect anything that is labeled
    # as a named entity.

    for subtree in tree:
        if hasattr(subtree, "label"):
            entity = " ".join(word for word, tag in subtree.leaves())
            named_entities.append(entity)

    return named_entities
```

```python
# This function creates trigrams from the cleaned tokens.
# A trigram is a group of 3 words that appear together.
# Counter is used to count how often each trigram appears,
# and most_common(10) returns the top 10 trigrams.
def get_trigrams(tokens):
    trigram_list = list(ngrams(tokens, 3))
    trigram_counts = Counter(trigram_list)
    return trigram_counts.most_common(10)
```

```python
results = {} # This dictionary will store all final results for each text.


# This loop goes through each text file and applies the same NLP process.
# For each text, it loads the file, preprocesses the words, counts word
# frequencies, finds named entities, and finds the most common trigrams.
for text_name, filename in texts.items():
    raw_text = load_text(filename)
    clean_tokens = preprocess_text(raw_text)

    freq_dist = FreqDist(clean_tokens)
    named_entities = count_named_entities(raw_text)
    common_trigrams = get_trigrams(clean_tokens)

        # Store all important results in one place so they can be printed later.
    results[text_name] = {
        "total_tokens": len(clean_tokens),
        "unique_tokens": len(set(clean_tokens)),
        "top_20_tokens": freq_dist.most_common(20),
        "named_entities": named_entities,
        "named_entity_count": len(named_entities),
        "top_10_trigrams": common_trigrams
    }
```

```python
# This section prints the results in a clean format for each text.
# It shows the total cleaned tokens, unique tokens, top 20 words,
# named entities, named entity count, and top 10 trigrams.

for text_name, data in results.items():
    print("=" * 70)
    print(text_name)
    print("=" * 70)

    print("Total tokens after filtering, stemming, and lemmatization:")
    print(data["total_tokens"])

    print("\nUnique tokens:")
    print(data["unique_tokens"])

    print("\nTop 20 tokens:")
    print(data["top_20_tokens"])

    print("\nNamed entity count:")
    print(data["named_entity_count"])

    print("\nNamed entities:")
    print(data["named_entities"])

    print("\nTop 10 trigrams:")
    print(data["top_10_trigrams"])
    print("\n")
```
